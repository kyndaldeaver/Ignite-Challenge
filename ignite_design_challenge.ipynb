```python

# calcualte reps percentage capacity
for rep in reps:
  if rep["Max_Capacity"] <= 0:
    rep["percentage_capacity"] = 1
  elif rep["Current_Capacity"] <= 0:
    rep["percentage_capacity"] = 0
  else:
    rep["percentage_capacity"] = rep["Current_Capacity"] / rep["Max_Capacity"]

# sort percentage capacity through ascending order
for _ in range(len(reps)):
  for i in range(len(reps)-1):
    if rep[i]["percentage_capacity"] > rep[i+1]["percentage_capacity"]:
      rep[i], rep[i+1] = rep[i+1], rep[i]
```

```python
leads_list = leads.to_dict(orient="records")

for lead in leads_list:
  # VIP override
  if lead["VIP_Referral"] == "true":
    lead["score"] = 100
    lead["priority_score"] = "critical"
    continue
  # starts everyone at 0
  score = 0

  # company size scoring (if no data, +20)
  size = lead["Company_Size"]
  if pd.notna(size) and str(size).strip() not in ["", "N/A", "nan", "None"]:
    comp_count = float(size)
    if comp_count >= 1000:
      score += 50
    elif comp_count >= 100:
      score += 30
    else:
      score += 0
  else:
    score += 20

  # recent activity scoring
  activity = lead["Recent_activity"]
  if activity in ["Requested Live Demo", "Contact Sales"]:
    score += 35
  elif activity in ["Attended Webinar", "Started Free Trial"]:
    score += 25
  elif activity == "Downloaded Whitepaper":
    score += 15
  # subcription shows interest, at least +5
  elif activity in ["Subscribed To Blog", "subscribed"]:
    score += 5
  elif activity == "Visited Pricing Page":
    score += 10
  else:
    score += 0

  # industry scoring
  vertical = lead["Industry_Vertical"]
  if vertical == "technology":
    score += 15
  elif vertical in ["healthcare", "finance"]:
    score += 10
  else:
    score += 0

  # sorting scores in groups
  lead["score"] = score
  if score >= 85:
    lead["priority_score"] = "critical"
  elif score >= 60:
    lead["priority_score"] = "high"
  elif score >= 40:
    lead["priority_score"] = "medium"
  else:
    lead["priority_score"] = "low"

# sort the leads scores in descending order
leads_list = sorted(leads_list, key=lambda x: x["score"], reverse=True)
```
